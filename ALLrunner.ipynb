# ALLrunner
Runs every `runner.ipynb` in each analysis folder sequentially.

```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
from pathlib import Path

# Folders containing a runner.ipynb, in execution order
runner_folders = [
    "Straight Run_hyeres",
    "Straight Run_hyeres_weight_analysis",
    "Straight Run_port_camargue",
    "Maneuvers_hyeres",
    "Maneuvers_port_camargue",
    "Starts_hyeres",
]

timeout = 1800  # seconds per runner

root = Path().resolve()

for folder in runner_folders:
    nb_path = root / folder / "runner.ipynb"
    print(f"\n=== Running {nb_path} ===")

    with nb_path.open(encoding="utf-8") as f:
        nb = nbformat.read(f, as_version=4)

    ep = ExecutePreprocessor(timeout=timeout, kernel_name="python3")

    try:
        ep.preprocess(nb, {"metadata": {"path": str(nb_path.parent)}})
        with nb_path.open("w", encoding="utf-8") as f_out:
            nbformat.write(nb, f_out)
        print(f"Done: {folder}")
    except Exception as e:
        print(f"ERROR in {folder}: {e}")
        raise

print("\nAll runners completed.")
```
