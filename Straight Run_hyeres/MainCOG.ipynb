# MainCOG – Straight Line Detection and Interval Extraction

This notebook processes raw sailing session data to automatically detect and extract the most relevant **straight line intervals** (upwind and downwind runs) for further analysis.

### Workflow
1. **Data loading**  
   - Scans all date folders and run subfolders inside `../Data_Sailnjord/Straight_lines` UPDATE: `../Data_Sailnjord/Port Camargue June 2025/Straight_lines`.  
   - Each run is expected to contain two CSV files (one per boat).  

2. **COG change detection**  
   - Uses a rolling mean approach to detect significant **Course Over Ground (COG)** changes.  
   - Identifies transition points between tacks/gybes and straight runs.  

3. **Trajectory and stability analysis**  
   - Plots the full trajectories of both boats.  
   - Computes intervals of stable sailing speed (SOG) between detected COG change points.  
   - Evaluates stability using smoothed SOG derivatives.  

4. **Interval selection**  
   - Extracts the **longest and most stable intervals** (by duration and stability score).  
   - Computes average performance metrics per interval (SOG, TWA, etc.).  

5. **Output**  
   - Saves results in `summary.json`, rebuilt from scratch at each run.  
   - Each entry contains:  
     - Run identifier  
     - Interval start and end times  
     - Duration  
     - Average SOG and TWA for both boats  
     - Stability metrics  

This notebook is the **first step of the analysis pipeline**, producing the core dataset of straight runs used in subsequent enrichment, merging, statistical analysis, and reporting.


```python
import os
import json
from cog_analysis import analyze_session
from natsort import natsorted
summary = []

# base_dir = "../Data_Sailnjord/Straight_lines"
base_dir = "../Data_Sailnjord/Hyères November 2025/Straight_lines"

# Scan de tous les dossiers de date
for date_folder in sorted(os.listdir(base_dir)):
    date_path = os.path.join(base_dir, date_folder)

    # Scan de tous les runs dans chaque dossier de date
    for run_folder in natsorted(os.listdir(date_path)):
        run_path = os.path.join(date_path, run_folder)

        # Cherche 2 fichiers .csv
        csv_files = [f for f in os.listdir(run_path) if f.endswith(".csv")]
        if len(csv_files) != 2:
            print(f"Ignoré: {run_path} (attendu 2 CSV, trouvé {len(csv_files)})")
            continue
        csv_paths = [os.path.join(run_path, f) for f in csv_files]
        print(f"Analyse: {csv_files[0]} et {csv_files[1]} dans {run_path}")
        try:
            longest_intervals = analyze_session(csv_paths[0], csv_paths[1])
            summary.append({
                "run": run_folder,
                "intervals": longest_intervals
            })
        except Exception as e:
            print(f"Erreur dans {run_path} : {e}")

# Sauvegarde du résumé (tout est reconstruit à chaque fois)
summary_file = "summary.json"
with open(summary_file, "w") as f:
    json.dump(summary, f, indent=2)

print(f"Résumé complet sauvegardé ({len(summary)} runs)")
```
